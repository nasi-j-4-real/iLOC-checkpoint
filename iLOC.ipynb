```python
import pandas as pd
```

```python
# 1. Use the iloc method to select the first 3 rows of the dataframe.
data = {
    'Name': ['John', 'Mary', 'Bob', 'Sarah', 'Tom', 'Lisa'], 
    'Department': ['IT', 'Marketing', 'Sales', 'IT', 'Finance', 'Marketing'], 
    'Age': [30, 40, 25, 35, 45, 28], 
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female'], 
    'Salary': [50000, 60000, 45000, 55000, 70000, 55000], 
    'Experience': [3, 7, 2, 5, 10, 4]
}
employee_df = pd.DataFrame(data)
```

```python
# 2. Use the loc method to select all rows where the Department is "Marketing".

first_three_rows = employee_df.iloc[0:3]

marketing_employees = employee_df.loc[employee_df['Department'] == 'Marketing']
```

```python
# 3.Use the iloc method to select the Age and Gender columns for the first 4 rows of the dataframe.
age_gender_first_four = employee_df.iloc[0:4, [2, 3]]

# Use the loc method to select the Salary and Experience columns for all rows where the Gender is "Male
male_salary_experience = employee_df.loc[employee_df['Gender'] == 'Male', ['Salary', 'Experience']]
```

```python
# 4. Code Verification 
print("--- First 3 Rows ---")
print(first_three_rows, "\n")

print("--- Marketing Department ---")
print(marketing_employees, "\n")

print("--- Age & Gender (First 4 Rows) ---")
print(age_gender_first_four, "\n")

print("--- Male Salary & Experience ---")
print(male_salary_experience)
```
